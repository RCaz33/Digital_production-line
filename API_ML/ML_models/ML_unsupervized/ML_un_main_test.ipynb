```python
import mlflow
from mlflow.entities import ViewType

from dotenv import load_dotenv
import os
load_dotenv()
```

```python
date = '2025-02-28'

user = os.getenv('LOCAL_DB_USER')
password = os.getenv('LOCAL_DB_PASS')
db = os.getenv('LOCAL_DB_MLFLOW')

mlflow.set_tracking_uri(f'mysql+pymysql://{user}:{password}@localhost/{db}')
```

```python
client = mlflow.MlflowClient()

experiment = client.get_experiment_by_name(f"/hdbscan_2025-02-28")

best_run = client.search_runs(
    experiment_ids=experiment.experiment_id,
    run_view_type=ViewType.ACTIVE_ONLY,
    max_results=5,
    order_by=["metrics.silhouette_score_val ASC"]
)[0]
```

```python
best_run.info.artifact_uri
```

```python
model_uri = f"{best_run.info.artifact_uri}/model"
model = mlflow.pyfunc.load_model(model_uri)
```
