```python
import requests
import pandas as pd

```

```python
url = 'http://127.0.0.1:8000'
headers = {
'accept': 'application/json',
'Content-Type': 'application/json'}
dataset='OGD'
```

```python
response = requests.get(f'{url}/{dataset}/', headers=headers)
metadata = pd.DataFrame(response.json())
```

```python
metadata.loc[:,'Batch_OGD_date'] = pd.to_datetime(metadata['Batch_OGD_date'].copy())
metadata.loc[metadata['Batch_OGD_date'] < pd.to_datetime('2025-01-01')]
```
